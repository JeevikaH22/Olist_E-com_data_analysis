```python
import unicodedata, re

def normalize_city(s):
    s = str(s).strip().lower()
    s = unicodedata.normalize("NFKD", s).encode("ascii", "ignore").decode()
    s = re.sub(r"[^a-z\s]", " ", s)      # drop digits, punctuation, odd chars
    s = re.sub(r"\s+", " ", s).strip()
    return s

print("before:", geoloc["geolocation_city"].nunique())
geoloc["city_clean"] = geoloc["geolocation_city"].apply(normalize_city)
print("after:", geoloc["city_clean"].nunique())
```

```python
geoloc["geolocation_state"] = geoloc["geolocation_state"].str.strip().str.upper()
geoloc["geolocation_state"].value_counts()
```

```python
def remove_accents(text):
    return ''.join(
        char for char in unicodedata.normalize('NFKD', text)
        if not unicodedata.combining(char))
```

```python
geoloc['city_normalized'] = (geoloc['geolocation_city'].str.lower().str.strip().apply(remove_accents))
```

```python
city_variations = (geoloc.groupby('city_normalized')['geolocation_city'].nunique().reset_index(name='variations'))
city_variations = city_variations[city_variations['variations'] > 1]
city_variations
```

```python
geoloc['geolocation_city'] = (geoloc['geolocation_city'].str.lower().str.strip().apply(remove_accents))
```

```python
geoloc.drop(columns=['city_normalized'], inplace=True)
```

```python
# drop helper columns left over from the outlier checks
geoloc = geoloc.drop(columns=["s_lat","s_lng","dist_state_km","limit",
                              "dist_zip_km","city_clean","geolocation_lat","geolocation_lng"],
                     errors="ignore")

# one row per zip prefix: most common city and state
geo_zip = (geoloc.groupby("geolocation_zip_code_prefix")
           .agg(geolocation_city=("geolocation_city", lambda s: s.mode().iloc[0]),
                geolocation_state=("geolocation_state", lambda s: s.mode().iloc[0]))
           .reset_index())

for df, col in [(geo_zip,"geolocation_zip_code_prefix"),
                (customers,"customer_zip_code_prefix"),
                (seller,"seller_zip_code_prefix")]:
    df[col] = df[col].astype(str).str.zfill(5)

assert geo_zip["geolocation_zip_code_prefix"].is_unique
print("zip rows:", len(geo_zip))
print("customers with no zip match:",
      (~customers["customer_zip_code_prefix"].isin(geo_zip["geolocation_zip_code_prefix"])).sum())
```

```python
def orphans(child, ccol, parent, pcol, label):
    print(f"{label:35s} orphan rows:", (~child[ccol].isin(parent[pcol])).sum())

orphans(orders, "customer_id", customers, "customer_id", "orders -> customers")
orphans(order_item, "order_id", orders, "order_id", "order_items -> orders")
orphans(order_item, "product_id", products, "product_id", "order_items -> products")
orphans(order_item, "seller_id", seller, "seller_id", "order_items -> sellers")
orphans(order_payment, "order_id", orders, "order_id", "order_payments -> orders")
orphans(order_review, "order_id", orders, "order_id", "order_reviews -> orders")

print("orders with no items   :", (~orders["order_id"].isin(order_item["order_id"])).sum())
print("orders with no payments:", (~orders["order_id"].isin(order_payment["order_id"])).sum())
print("orders with no review  :", (~orders["order_id"].isin(order_review["order_id"])).sum())
```

```python
def clean_text(s):
    s = remove_accents(str(s).lower().strip())
    return " ".join(s.split())

for df, city_col, state_col in [(seller,"seller_city","seller_state"),
                                (customers,"customer_city","customer_state")]:
    df[city_col] = df[city_col].apply(clean_text)
    df[state_col] = df[state_col].str.strip().str.upper()

junk = seller[seller["seller_city"].str.contains(r"[\d@/\\,]", regex=True)]
print("junk-looking seller cities:", len(junk))
junk[["seller_id","seller_city","seller_state"]]
```

```python
manual = {"sbc": "sao bernardo do campo", "sp": "sao paulo",
          "s jose do rio preto": "sao jose do rio preto"}

def tidy_city(c):
    c = re.split(r"[/\\,(]| - ", c)[0].strip()                              # cut at / \ , ( or " - "
    c = re.sub(r"[-\s]+(sp|rj|sc|pr|df|mg|rs|es|ba|go|pe|ce)$", "", c)      # trailing state code
    return manual.get(c, c)

valid = set(geoloc["geolocation_city"])
zip_city  = geo_zip.set_index("geolocation_zip_code_prefix")["geolocation_city"]
zip_state = geo_zip.set_index("geolocation_zip_code_prefix")["geolocation_state"]

# 1) tidy patterns, only on cities that are not already valid
bad = ~seller["seller_city"].isin(valid)
seller.loc[bad, "seller_city"] = seller.loc[bad, "seller_city"].apply(tidy_city)

# 2) still not valid (typos, 'bahia', 'centro', emails, digits) -> take city from zip prefix
bad = ~seller["seller_city"].isin(valid)
review = seller.loc[bad, ["seller_city", "seller_state", "seller_zip_code_prefix"]].copy()
review["city_from_zip"] = review["seller_zip_code_prefix"].map(zip_city)
seller.loc[bad, "seller_city"] = review["city_from_zip"].fillna(review["seller_city"])
print("fixed via zip:", len(review))
print(review.to_string())                       # read this table

# 3) state that disagrees with the zip's state
zs = seller["seller_zip_code_prefix"].map(zip_state)
mism = zs.notna() & (zs != seller["seller_state"])
print("state != zip state:", mism.sum())
print(seller.loc[mism, ["seller_city", "seller_state", "seller_zip_code_prefix"]].assign(zip_state=zs[mism]))
seller.loc[mism, "seller_state"] = zs[mism]

# 4) what is still unmatched
left = seller[~seller["seller_city"].isin(valid)]
print("still not in geolocation:", len(left))
left[["seller_id", "seller_city", "seller_state", "seller_zip_code_prefix"]]
```
