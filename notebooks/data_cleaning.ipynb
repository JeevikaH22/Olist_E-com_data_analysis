```python
import unicodedata, re

def normalize_city(s):
    s = str(s).strip().lower()
    s = unicodedata.normalize("NFKD", s).encode("ascii", "ignore").decode()
    s = re.sub(r"[^a-z\s]", " ", s)      # drop digits, punctuation, odd chars
    s = re.sub(r"\s+", " ", s).strip()
    return s

print("before:", geoloc["geolocation_city"].nunique())
geoloc["city_clean"] = geoloc["geolocation_city"].apply(normalize_city)
print("after:", geoloc["city_clean"].nunique())
```

```python
geoloc["geolocation_state"] = geoloc["geolocation_state"].str.strip().str.upper()
geoloc["geolocation_state"].value_counts()
```

```python
geoloc['geolocation_city'].nunique()

sorted(geoloc['geolocation_city'].unique())
```

```python
def remove_accents(text):
    return ''.join(
        char for char in unicodedata.normalize('NFKD', text)
        if not unicodedata.combining(char))
```

```python
geoloc['city_normalized'] = (geoloc['geolocation_city'].str.lower().str.strip().apply(remove_accents))
```

```python
city_variations = (geoloc.groupby('city_normalized')['geolocation_city'].nunique().reset_index(name='variations'))

city_variations = city_variations[city_variations['variations'] > 1]

city_variations
```

```python
for city in city_variations['city_normalized']:
    print(city,":",geoloc.loc[geoloc['city_normalized'] == city,'geolocation_city'].unique())
```

```python
geoloc['geolocation_city'] = (geoloc['geolocation_city'].str.lower().str.strip().apply(remove_accents))
```

```python
geoloc.drop(columns=['city_normalized'], inplace=True)
```

```python
geoloc["geolocation_zip_code_prefix"].count()
```

```python
lat, lng = geoloc["geolocation_lat"], geoloc["geolocation_lng"]
outside = ~(lat.between(-33.8, 5.3) & lng.between(-74.0, -34.7))
print("outside Brazil:", outside.sum())

geoloc[outside].head(10) 
geoloc = geoloc[~outside].copy()
```

```python
def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    p1, p2 = np.radians(lat1), np.radians(lat2)
    dphi = p2 - p1
    dl = np.radians(lon2 - lon1)
    a = np.sin(dphi/2)**2 + np.cos(p1)*np.cos(p2)*np.sin(dl/2)**2
    return 2 * R * np.arcsin(np.sqrt(a))

state_med = (geoloc.groupby("geolocation_state")[["geolocation_lat","geolocation_lng"]]
               .median().rename(columns={"geolocation_lat":"s_lat","geolocation_lng":"s_lng"}))
geoloc = geoloc.join(state_med, on="geolocation_state")
geoloc["dist_state_km"] = haversine(geoloc["geolocation_lat"], geoloc["geolocation_lng"],
                                 geoloc["s_lat"], geoloc["s_lng"])

# robust per-state threshold (IQR rule), since states differ a lot in size
q = geoloc.groupby("geolocation_state")["dist_state_km"].quantile([0.25, 0.75]).unstack()
q["limit"] = q[0.75] + 3 * (q[0.75] - q[0.25])
geoloc = geoloc.join(q["limit"], on="geolocation_state")

wrong_state = geoloc["dist_state_km"] > geoloc["limit"]
print("far from own state:", wrong_state.sum())
geoloc[wrong_state][["geolocation_zip_code_prefix","geolocation_city",
                  "geolocation_state","dist_state_km","limit"]].head(15)
```

```python
zip_med = geoloc.groupby("geolocation_zip_code_prefix")[["geolocation_lat","geolocation_lng"]].transform("median")
geoloc["dist_zip_km"] = haversine(geoloc["geolocation_lat"], geoloc["geolocation_lng"],
                               zip_med["geolocation_lat"], zip_med["geolocation_lng"])
geoloc["dist_zip_km"].describe(percentiles=[.9, .99, .999])
```
